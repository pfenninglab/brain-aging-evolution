```r
# Manuscript Supp Figure 9 --------------------------------------------------------
#----------------------------------------------------------------------------------

# phyloP scores -- VGP 577
library(tidyverse)
library(ggpubr)   # for significance brackets

DATADIR <- "/Aging/LQ_TACIT/data/"

# Load
test <- read_tsv(paste0(DATADIR, "phylop_out/Overlap_all_nuclei_neg_aging_pos_output.tsv"),
                 col_names = c("name","size","covered","sum","mean0","mean"))
ctrl <- read_tsv(paste0(DATADIR, "phylop_out/all_nuclei_lq_neg_427_random_output.tsv"),
                 col_names = c("name","size","covered","sum","mean0","mean"))
ctrl2 <- read_tsv(paste0(DATADIR, "phylop_out/all_nuclei_lq_neg_427_random2_output.tsv"),
                  col_names = c("name","size","covered","sum","mean0","mean"))

# Mann-Whitney (one-sided: test > control)
wilcox.test(test$mean, ctrl$mean,  alternative = "greater")
wilcox.test(test$mean, ctrl2$mean, alternative = "greater")

# ---- Assemble data ----
plot_df <- bind_rows(
  test  %>% mutate(group = "vulnerable"),
  ctrl  %>% mutate(group = "random 1"),
  ctrl2 %>% mutate(group = "random 2")
) %>%
  drop_na(mean) %>%
  mutate(group = factor(group, levels = c("vulnerable",
                                          "random 1",
                                          "random 2")))

pal <- c("vulnerable" = "#D85A30",
         "random 1" = "#1D9E75",
         "random 2"    = "#378ADD")

# n labels placed just below the lowest value
n_labels <- plot_df %>%
  group_by(group) %>%
  summarise(n = n(), y = min(mean), .groups = "drop")

ymin <- min(plot_df$mean); ymax <- max(plot_df$mean)

# ---- Plot ----
p <- ggplot(plot_df, aes(x = group, y = mean, fill = group)) +
  geom_hline(yintercept = 0, linetype = "dashed",
             color = "grey60", linewidth = 0.4) +
  # violin: distribution shape
  geom_violin(aes(color = group), alpha = 0.18,
              linewidth = 0.7, width = 0.9,
              trim = FALSE, scale = "width") +
  # narrow boxplot inside, with a clear white-filled median
  geom_boxplot(width = 0.12, alpha = 0.9, outlier.shape = NA,
               color = "grey25", linewidth = 0.5,
               show.legend = FALSE) +
  stat_summary(fun = median, geom = "point",
               shape = 21, size = 2,
               fill = "white", color = "grey25") +
  # sample sizes
  geom_text(data = n_labels,
            aes(x = group, y = y, label = paste0("n = ", n)),
            vjust = 1.8, size = 3.4, color = "grey30",
            inherit.aes = FALSE) +
  scale_fill_manual(values = pal) +
  scale_color_manual(values = pal) +
  scale_y_continuous(expand = expansion(mult = c(0.08, 0.12))) +
  labs(x = NULL,
       y = "Mean phyloP score\n(VGP 577-way ssrev)") +
  theme_classic(base_size = 13) +
  theme(
    legend.position   = "none",
    axis.title.y      = element_text(margin = margin(r = 8), face = "plain"),
    axis.text.x       = element_text(size = 11, color = "black", lineheight = 0.9),
    axis.text.y       = element_text(size = 11, color = "black"),
    axis.line         = element_line(linewidth = 0.4),
    axis.ticks        = element_line(linewidth = 0.4),
    plot.margin       = margin(12, 16, 12, 12)
  )

# ---- Add significance brackets (one-sided Wilcoxon) ----
p <- p +
  stat_compare_means(
    comparisons = list(c("vulnerable", "random 1"),
                       c("vulnerable", "random 2")),
    method = "wilcox.test",
    method.args = list(alternative = "greater"),
    label = "p.signif",          # use "p.format" for numeric p-values
    tip.length = 0.01,
    size = 4,
    color = "grey25"
  )

p

ggsave(paste0(PLOTDIR, "phyloP_427_vuln_random.pdf"),
       plot = p, width = 4, height = 4, device = cairo_pdf)
```
